```python

import matplotlib.pyplot as plt

archivo = "../out/dosis_region3.out"

with open(archivo, "r") as f:
    for linea in f:
        datos = linea.split()
        if len(datos) >= 5 and datos[0].isdigit() and datos[1] == "700" and "#" not in datos:
            dosis = float(datos[3])
            error = float(datos[4])
            break

plt.figure(figsize=(6, 4))

plt.bar("Región 700", dosis, color="steelblue", width=0.4)

plt.text(0, dosis * 1.04,
         f"{dosis:.4f} Gy/source\nError: {error*100:.2f}%",
         ha="center", fontsize=11)

plt.ylabel("Dosis [Gy/source]")
plt.title("Dosis depositada por protones")
plt.ylim(0, dosis * 1.2)
plt.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()
```
